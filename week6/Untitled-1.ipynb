```python
# ============================================
# WORLD CUP PLAYERS HISTORICAL DATASET
# Exporta jugadores de todos los mundiales a Excel
# ============================================

# Requisitos:
# pip install pandas requests beautifulsoup4 lxml openpyxl

import pandas as pd
import requests
from bs4 import BeautifulSoup
import time

BASE_URL = "https://en.wikipedia.org/wiki/"

# Mundiales desde 1930
WORLD_CUPS = [
    1930, 1934, 1938,
    1950, 1954, 1958,
    1962, 1966, 1970,
    1974, 1978, 1982,
    1986, 1990, 1994,
    1998, 2002, 2006,
    2010, 2014, 2018,
    2022,2026
]

headers = {
    "User-Agent": "Mozilla/5.0"
}

all_players = []

def get_squads(year):
    """
    Obtiene planteles oficiales desde Wikipedia.
    """
    url = f"{BASE_URL}{year}_FIFA_World_Cup_squads"

    print(f"Procesando {year}...")

    response = requests.get(url, headers=headers)

    if response.status_code != 200:
        print(f"Error en {year}")
        return

    soup = BeautifulSoup(response.text, "lxml")

    # Cada selección suele venir como h2/h3
    current_team = None

    for tag in soup.find_all(["h2", "h3", "table"]):

        # Detectar selección
        if tag.name in ["h2", "h3"]:
            headline = tag.get_text(strip=True)

            # Limpiar texto
            headline = headline.replace("[edit]", "")

            if len(headline) < 40:
                current_team = headline

        # Detectar tablas de jugadores
        if tag.name == "table":
            rows = tag.find_all("tr")

            for row in rows[1:]:

                cols = row.find_all(["td", "th"])

                if len(cols) < 2:
                    continue

                row_data = [c.get_text(" ", strip=True) for c in cols]

                # Buscar nombre probable
                player_name = None

                for cell in row_data:
                    if len(cell) > 3 and not cell.isdigit():
                        player_name = cell
                        break

                if not player_name:
                    continue

                record = {
                    "world_cup": year,
                    "country": current_team,
                    "player": player_name,
                    "raw_data": " | ".join(row_data)
                }

                all_players.append(record)

    time.sleep(1)


# Ejecutar scraping
for wc in WORLD_CUPS:
    try:
        get_squads(wc)
    except Exception as e:
        print(f"Error en {wc}: {e}")


# Crear DataFrame
df = pd.DataFrame(all_players)

# Limpieza básica
df = df.drop_duplicates()

# Guardar Excel
output_file = "world_cup_players_history.xlsx"

df.to_excel(output_file, index=False)

print("\n===================================")
print(f"Archivo generado: {output_file}")
print(f"Registros totales: {len(df)}")
print("===================================")
```
